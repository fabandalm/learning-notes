# NumPy Random Numbers

```python
# random decimal numbers
res = np.random.rand(4,2)
res
```

```python
res[0]
```

```python
initial_centroids = np.array([[1,2], [6,2], [8,5]])
```

```python
x = res[0] - initial_centroids[0]
x
```

```python
norm_ij = np.linalg.norm(x, ord=3)
norm_ij
```

```python
a = abs(x[0])**2
a
```

```python
b = abs(x[1])**2
b
```

```python
res = a + b
res
```

```python
import math
math.sqrt(res)
```

# NumPy Copy Array

```python
a = np.array([1,2,3])
b = a
b[0] = 100

print(a)
```

```python
a = np.array([1,2,3])
b = a.copy()
b[0] = 100

print(a)
```

# Reorganizing Arrays

```python
before = np.array([[1,2,3,4],[5,6,7,8]])
print(before)

print("###")
# after = before.reshape((8,1))
after = before.reshape((2,2,2))
print(after)
```

```python
v1 = np.array([1,2,3,4])
v2 = np.array([5,6,7,8])

np.vstack([v1,v2])
```

```python
h1 = np.ones((2,4))
h1
h2 = np.zeros((2,2))
h2
np.hstack((h1,h2))
```
