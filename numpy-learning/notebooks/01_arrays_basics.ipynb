```python
import numpy as np
import sys
```

# The Basics

```python
a = np.array([1,2,3], dtype='int8')
b = np.array([[9.0,8.0,7.0],[6.0,5.0,4.0]])

# Print Matrix
print(a)
```

```python
# Get Dimension
print("Dimension", a.ndim)
```

```python
# Get Type/Size
print("Type", a.dtype)
print("Size", a.itemsize)
```

```python
# Get Type/Size
print("Type", b.dtype) # data type
print("Size", b.itemsize) # number of items
```
